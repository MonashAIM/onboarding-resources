<h2> <span style="color: #f28f2c;"> Part 3. Supervised vs Unsupervised vs Reinforcement Learning </span> </h2>

We will be discussing three different types of Machine Learning: **supervised learning**, **unsupervised learning**, and **reinforcement learning**.


Throughout this section, we will use a **fruit basket example** containing apples, bananas, and oranges to help illustrate how each learning approach differs.
 

#### **Supervised Learning**
In supervised learning, the algorithm learns from **labelled data**, where each input (such as colour, shape, or size) is paired with a known output (such as “apple”, “banana”, or “orange”).  
For example, if you show a computer labelled examples of fruits, it can learn the relationship between their features and labels. Once trained, it can correctly identify new, unseen fruits based on these learned patterns.  

The goal of supervised learning is to learn a **mapping function** that can accurately predict the output (y) for new, unseen data (x).

#### **Unsupervised Learning**
In unsupervised learning, the dataset has **no labels**. The algorithm explores the data to discover **hidden patterns or groupings** without explicit guidance.  
Using the same fruit basket example, the model would examine characteristics like colour or shape and might group similar fruits together, such as clustering round fruits (apples and oranges) separately from long curved fruits (bananas).  

The goal is to find inherent **patterns, relationships, or structures** within the data.

#### **Reinforcement Learning**
In reinforcement learning, an **agent interacts with an environment** by taking actions and receiving **feedback** in the form of rewards or penalties.  
For example, imagine a robotic arm learning to pick ripe fruits. It gains rewards for selecting ripe fruits and penalties for unripe ones. Over time, it learns through **trial and error** which actions lead to the highest cumulative reward.  

Unlike supervised or unsupervised learning, reinforcement learning **does not use labelled datasets** - it learns from experience and feedback to make better decisions over time.

<h2> <span style="color: #f28f2c;"> Part 4. Data Preparation & Cleaning </span> </h2>

Before training any model, it’s important to ensure your dataset is **accurate, consistent, and well-structured**!
Even the most advanced algorithm cannot perform well if the data is messy or misleading. In machine learning, **data cleaning** is the process of preparing raw data so that models can learn meaningful patterns instead of noise or errors.

A well-cleaned dataset allows the model to:
- Focus on **relevant patterns** rather than artefacts or mistakes  
- Produce **more reliable and fair predictions**  
- Reduce training time by avoiding unnecessary complexity

Think of it as "Garbage In, Garbage Out"! 🗑️ The quality of your models heavily depend on the quality of your dataset.   

#### **Common Data Issues**

When working with real-world data, you’ll often encounter:
- **Missing values** (e.g., incomplete survey responses or broken sensors)  
- **Duplicates** (repeated records that skew results)  
- **Inconsistent formats** (e.g., “Yes/No” vs. “Y/N”)  
- **Outliers or extreme values** (unusual data points that distort averages)  
- **Incorrect data types** (e.g., numbers stored as text)




<h2> <span style="color: #f28f2c;"> Part 5. Linear Regression </span> </h2>

Linear Regression is one of the most fundamental algorithms in Machine Learning.
It is used when we want to **predict a continuous value** based on one or more input variables.

For example, we might want to predict a fruit’s **weight** from its **size**, or estimate a house’s **price** from its **area**.  
The idea is to find a **line of best fit** that captures the relationship between the input (`X`) and the output (`y`).

In its simplest form, a linear regression model can be written as:

\[
y = w x + $\beta$
\]

- **x** → input variable (e.g. fruit size)  
- **y** → predicted output (e.g. fruit weight)  
- **w** → weight or slope of the line (how much `y` changes with `x`)  
- **$\beta$** → bias or intercept (where the line crosses the y-axis)

Our goal is to find the values of **w** and **$\beta$** that make our predictions as close as possible to the actual data points.

#### **Method**

Linear regression works by finding the **best-fitting line** through the data points.  
This line represents the relationship between input features (X) and the target output (y).

To find this line, the model tries different values for the parameters **w** (weight/slope) and **b** (bias/intercept) and measures how well each combination fits the data.  
It does this through an **iterative process**, where it repeatedly updates the parameters step by step to reduce prediction errors.  


![Linear Regression Gif](https://gbhat.com/assets/gifs/linear_regression.gif)

You can use **Linear Regression** when:
- The relationship between input(s) and output is **approximately linear** (a straight-line trend).
- The target variable is **continuous** (e.g. predicting weight, price, or temperature).
- The features are **not highly correlated** with each other (to avoid redundancy or instability).

If the relationship is curved, highly complex, or categorical, other models (like polynomial regression or logistic regression) may be more appropriate.


#### **Loss Functions**

A **loss function** measures how far off our model’s predictions are from the actual data. Basically calculating *how bad* our current model is :'(
There are many different loss functions including Mean Squared Error (MSE), Least Squared Error (LSE), Sum of Residuals.

#### **Gradient Descent**
How do we relate this loss function back to our weights and biases parameters? CALCULUS!😔   

**Gradient Descent** is an optimisation algorithm that helps the model find the best parameters (**w** and **b**) by gradually reducing the loss.

Think of it like climbing down a hill:  
- Each step takes you in the direction that **reduces error** the most.  
- The **slope** of the loss function tells you which direction to move (that’s the "gradient").  
- The **learning rate** controls how big each step is. Too big and you might overshoot! Too small and it takes too long...

Over many iterations, the model "descends" to the lowest point of the loss curve, this is where the loss is smallest, and the model performs best.

#### **Example**

**Visualising the Data**

Let’s start by plotting a small dataset.  
We’ll generate some random data to simulate a simple linear relationship, such as fruit size vs. weight.


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate sample data
np.random.seed(0)
X = 2 * np.random.rand(100, 1)
y = 4 + 3 * X + np.random.randn(100, 1)

plt.scatter(X, y)
plt.title("Example Data: Relationship between X and y")
plt.xlabel("X")
plt.ylabel("y")
plt.show()
```

**Training a Linear Regression Model**

Now that we have our dataset, we can train a **Linear Regression** model.  
The model will learn the best values of **w** (weight) and **b** (bias) that fit the data.   
***NEED TO UPDATE TO USE METHODS IN WORKSHOP***


```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X, y)

# Display learned parameters
print(f"Weight (w): {model.coef_[0][0]:.2f}")
print(f"Bias (b): {model.intercept_[0]:.2f}")
```

**Visualising the Fitted Line**

Let’s see how well our model fits the data.  
We’ll plot the original points and the predicted regression line.


```python
y_pred = model.predict(X)

plt.scatter(X, y, label="Actual data")
plt.plot(X, y_pred, color="red", label="Regression line")
plt.legend()
plt.title("Fitted Linear Regression Model")
plt.show()
```
