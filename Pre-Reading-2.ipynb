<h2><span style="color: #f28f2c;">2. What is a neural network?</span></h2>

Moving on from our linear regression and classification models, the focus of this workshop will be on the new model mentioned previously, <b>neural networks</b>. In fact, neural networks are flexible models that can be tailored to tasks such as classification (assigning inputs to discrete categories), regression (predicting temperature), and others depending on the type of dataset being fed into the model and the type of output information generated. However, the manner that a neural network operates, in comparison to what we have covered thus far, is slightly more complicated. 

The goal of a neural network is to generate and learn complex patterns in the data that it is presented with. This is achieved through <b>mapping multiple layers of neurons</b> from the input features to your final label, as shown in the figure below:  





Neurons are essentially small units with an associated weight and bias that perform computations. The weight of a neuron essentially acts like a faucet that controls how much influence a neuron will have on downstream elements. 

In <b>hidden layer 1</b>, we see that each of the neurons (blue) are receiving inputs from all three input neurons (yellow); however, the input neurons will be scaled based on their weight values depending on which input features are more significant in contributing to an accurate final prediction. This can be described as a <b>weighted sum of inputs</b> and is given by the following formula: 

$$
S = w_1 \cdot x_1 + w_2 \cdot x_2 + ... + w_n \cdot x_n
$$

This continues with our <b>second hidden layer</b> where each of the neurons are receiving weighted sums from the hidden layer 1 neurons: 

$$
T = w_{S1} \cdot S_1 + w_{S2} \cdot S_2 + ... w_{Sn} \cdot S_n
$$


This process continues until we arrive at our final output layer where we need to map the final layer of neurons to a final prediction. 

If you have a biology background, the neurons in a neural network function essentially in the same manner as biological neurons and in fact, was where the inspiration for neural networks originated from when Warren McCulloch and Walter Pitts first proposed the model in 1943, followed by Franken Rosenblatt in 1958 who improved the model to perform pattern recognition. Biological neurons receive inputs from multiple neurons at synaptic connections, integrating this information to determine whether they should fire an action potential (electrical signal). Each neuron will have a threshold for action potential firing: if this threshold is surpassed, then the neuron will fire an action potential; otherwise, the neuron will not fire an action potential. 

We discussed how assigning weights to each of the input features will determine their level of influence on the final weighted sum. Remember that we also have <b>bias terms</b> for each neuron which essentially set the threshold, determining if a neuron will fire an action potential. 

<b>Another analogy:</b> imagine that each of our input layer neurons (features) are lakes connected to dams. Each of these will run to more lakes (hidden layer 1 and hidden layer 2) before arriving at a final output source where we are trying to determine if the water is safe to drink. By adjusting the valves of each lake, we can effectively control how much water from that lake will flow into another lake downstream. 

Let's assume that we have randomly preset our valves and received some water that we determined to be non-safe to drink. Remember the main goal of a machine learning problem is <b>to learn and improve</b>. Hence, our goal is to determine which lakes have a greater ability to pollute the drinking water compared to other lakes and close them. These lakes will be assigned a greater weight value as they will heavily influence whether we have safe or non-safe drinking water. 

<h2><span style="color: #f28f2c;">3. Components of a neural network</span></h2>

The figure above is essentially summarises the main components of our neural network: 

1. First, we have a layer of input features. 
2. We feed weighted sums from the previous layer into the neurons in our next layer (denoted by the $\sum$ symbol) 
3. We also notice that these neurons also perform a function (denoted by the $f$ symbol), known as an activation function which will be covered shortly
4. This first hidden layer is fed into a second hidden layer in a similar manner (weighted sums followed by an activation function)
5. Finally, this is mapped back to the final output layer (prediction) which in this case, is a multivariate classification problem where we are trying to predict multiple things

<h3>Hidden layers and activation functions</h3>

We won't delve into the complexities here so for now, just remember that hidden layers and activation functions are two components of our neural network architecture. 
- <b>Hidden layers:</b> having more hidden layers allows the model to extract and learn increasingly complex features and patterns (at higher computational loads)
- <b>Activation functions:</b> without activation functions between 2 hidden layers, these 2 hidden layers will be seen as 1 by the model, removing the degree of complexity that we tried to implement into our model 

<h2><span style="color: #f28f2c;">4. The training workflow </span></h2>

The main process in training a neural network can be divided into three main phases: forward pass, backpropagation, and optimization. In fact, all three have been covered to some degree already in our first workshop, however, there are a few complexities that complicate this process as a result of our new neural network model. 

<h3>For now, we want you to be able to identify these three stages, but here's a brief sneak peek into what they encompass: </h3>

<h4><b>Forward pass:</b></h4>

- Involves feeding our input data through each layer of the neural network to arrive at a final prediction. 
- Remember that we achieve this by calculating the weighted sum of each neuron, passing through activation functions if we have multiple hidden layers, until we arrive at a final output. 
- Forward pass also encompasses our loss function, where we reflect on how accurate our prediction is.

<h4><b>Backpropagation: </b></h4>

- This process essentially tries to do the same thing before where we use calculus to determine the gradient of the loss function with respect to the weights and biases. 
- The only difference is that now, we have multiple layers of neurons that are influenced by previous neurons. 
- Hence, instead of calculating a single gradient, we calculate the gradients for each layer and combine these using the chain rule into a single derivative, 

<h4><b>Optimization: </b></h4>

- Calculating this gradient informs us which direction we should update our weights/biases to minimise the loss. 
- Hence, we can use our gradient descent equation to update the weights and biases of our model before repeating this workflow for a desired number of iterations.
